5. Resumos e KPIs

```python
dados.groupby("categoria")["custo_total"].sum() #primeira linha do relatorio
```

```python
#agg serve para agrupar mais de um
resumo_categoria= dados.groupby("categoria").agg(  #estrutura do .agg(): nome_da_coluna_nova=("coluna_de_onde_vem","que_conta_devo_fazer")
    chamados=("id_chamado","count"),
    horas_totais=("horas_gastas","sum"),
    custo_total=("custo_total","sum")
)
resumo_categoria
```

```python
dados["mes"] = dados["data_abertura"].dt.strftime("%Y-%m")
dados["mes"].value_counts()
```

```python
resumo_mes = dados.groupby("mes").agg(
    chamados     = ("id_chamado",   "count"),
    horas_totais = ("horas_gastas", "sum"),
    custo_total  = ("custo_total",  "sum"),
)
resumo_mes
```

```python
dados.groupby("prioridade")["dias_para_resolver"].mean()
resumo_prioridade = dados.groupby("prioridade").agg(
    chamados   = ("id_chamado",         "count"),
    dias_medio = ("dias_para_resolver", "mean"),
)
resumo_prioridade
```
